```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import warnings

# Silenciamos advertencias de NumPy por si el algoritmo intenta evaluar logaritmos de números negativos
warnings.filterwarnings('ignore', category=RuntimeWarning)

# =====================================================================
# 1. Definición de Funciones y sus Derivadas usando SYMPY
# =====================================================================
x_sym = sp.Symbol('x')

# Declaramos las expresiones matemáticas simbólicas
f1_sym = sp.cos(x_sym) + x_sym**2
f2_sym = sp.exp(-x_sym) * (x_sym**2 + 5*x_sym)
f3_sym = sp.log(x_sym + 1) + x_sym**2
f4_sym = x_sym**4 - 5*x_sym**2 + 4*x_sym

# SymPy calcula las derivadas automáticamente con sp.diff()
df1_sym = sp.diff(f1_sym, x_sym)
df2_sym = sp.diff(f2_sym, x_sym)
df3_sym = sp.diff(f3_sym, x_sym)
df4_sym = sp.diff(f4_sym, x_sym)

# sp.lambdify() convierte las expresiones simbólicas en funciones ejecutables por NumPy
h1 = sp.lambdify(x_sym, f1_sym, 'numpy')
dh1 = sp.lambdify(x_sym, df1_sym, 'numpy')

h2 = sp.lambdify(x_sym, f2_sym, 'numpy')
dh2 = sp.lambdify(x_sym, df2_sym, 'numpy')

h3 = sp.lambdify(x_sym, f3_sym, 'numpy')
dh3 = sp.lambdify(x_sym, df3_sym, 'numpy')

h4 = sp.lambdify(x_sym, f4_sym, 'numpy')
dh4 = sp.lambdify(x_sym, df4_sym, 'numpy')

# =====================================================================
# 2. Algoritmo de Descenso del Gradiente
# =====================================================================
def gradiente_descendente(f, df, x0, learning_rate, max_epochs=100, tol=1e-6):
    x_path = [x0]
    x = x0
    iteraciones_reales = 0

    for i in range(max_epochs):
        grad = df(x)
        x_nuevo = x - learning_rate * grad

        # Prevenir desbordamientos numéricos si la tasa de aprendizaje es muy alta y diverge
        if np.isnan(x_nuevo) or np.isinf(x_nuevo) or abs(x_nuevo) > 1e4:
            break

        x_path.append(x_nuevo)
        iteraciones_reales += 1

        if abs(x_nuevo - x) < tol:
            break

        x = x_nuevo

    return np.array(x_path), iteraciones_reales

# =====================================================================
# 3. Configuración de Experimentos
# =====================================================================
funciones_test = [
    {
        "name": "h(x) = cos(x) + x^2",
        # Usamos str() para que SymPy nos imprima la derivada que calculó
        "df_name": str(df1_sym),
        "f": h1, "df": dh1,
        "x0_list": [-2.0, 0.5, 2.5],
        "lr_list": [0.01, 0.1, 0.8],
        "x_range": (-3, 3)
    },
    {
        "name": "h(x) = e^{-x}(x^2 + 5x)",
        "df_name": str(df2_sym),
        "f": h2, "df": dh2,
        "x0_list": [0.0, 2.0, 6.0],
        "lr_list": [0.05, 0.2, 0.8],
        "x_range": (-2, 8)
    },
    {
        "name": "h(x) = ln(x + 1) + x^2",
        "df_name": str(df3_sym),
        "f": h3, "df": dh3,
        "x0_list": [-0.5, 0.5, 1.5],
        "lr_list": [0.05, 0.1, 0.4],
        "x_range": (-0.99, 2)
    },
    {
        "name": "h(x) = x^4 - 5x^2 + 4x",
        "df_name": str(df4_sym),
        "f": h4, "df": dh4,
        "x0_list": [-2.5, 0.0, 2.5],
        "lr_list": [0.01, 0.05, 0.09],
        "x_range": (-3, 3)
    }
]

# =====================================================================
# 4. Bucle Principal de Aplicación y Visualización
# =====================================================================
for func in funciones_test:
    print(f"\n" + "="*80)
    print(f" ANÁLISIS DE LA FUNCIÓN: {func['name']}")
    print(f" Derivada (calculada por SymPy): {func['df_name']}")
    print("="*80)

    fig, axs = plt.subplots(3, 3, figsize=(15, 12))
    fig.suptitle(f"{func['name']}", fontsize=18, fontweight='bold', y=0.98)

    x_vals = np.linspace(func["x_range"][0], func["x_range"][1], 200)
    y_vals = func["f"](x_vals)

    for i, x0 in enumerate(func["x0_list"]):
        for j, lr in enumerate(func["lr_list"]):

            # Ejecutar Descenso del Gradiente
            path_x, iter_count = gradiente_descendente(func["f"], func["df"], x0, lr, max_epochs=100)
            path_y = func["f"](path_x)

            minimo_x = path_x[-1]
            minimo_y = path_y[-1]

            # Imprimir resultados del escenario
            print(f" Inicio: x0={x0:5.1f} | LR={lr:4.2f} -> Iteraciones: {iter_count:3d} | Mínimo en x={minimo_x:7.4f}, h(x)={minimo_y:7.4f}")

            # Graficar
            ax = axs[i, j]
            ax.plot(x_vals, y_vals, color="blue", linewidth=1.5, alpha=0.6)
            ax.plot(path_x, path_y, 'ro-', markersize=4, alpha=0.7)
            ax.plot(path_x[0], path_y[0], 'go', markersize=6)
            ax.plot(minimo_x, minimo_y, 'k*', markersize=10)

            ax.set_title(f"x0={x0} | lr={lr} | Iters={iter_count}", fontsize=10)
            ax.grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
```

El descenso del gradiente funciona mejor en funciones polinomiales debido a su suavidad, lo que garantiza una convergencia más estable hacia mínimos locales, aunque el resultado depende del punto de inicio. En cambio, las funciones no polinomiales presentan mayor dificultad por su alta sensibilidad a la tasa de aprendizaje, riesgos de desbordamiento, no convexidad y presencia de asíntotas o dominios restringidos que pueden causar divergencia.

Diferencias entre funciones polinomiales y no polinomiales en cuanto a la efectividad del algoritmo

1.	Considerando la función polinomial. h(x) = x^4 - 5x^2 + 4x:

    Convergencia: El algoritmo de descenso del gradiente tiende a ser bastante efectivo para funciones polinomiales. En el ejemplo, se observa que, para la mayoría de los puntos iniciales y tasas de aprendizaje, el algoritmo converge a un mínimo local en un número razonable de iteraciones.

    Las funciones polinomiales generalmente no tienen picos, tampoco cambios bruscos de dirección y son continuas, lo que facilita que el gradiente apunte consistentemente hacia un mínimo.

    Las funciones polinomiales de grado superior pueden tener varios mínimos locales. El algoritmo de descenso del gradiente, al ser un método local, converge al mínimo más cercano al punto de inicio. Como se observa en varios de los ejercicios.

2.	Considerando las funciones  No Polinomiales (ej. h(x) = cos(x) + x^2, h(x) = e^{-x}(x^2 + 5x), h(x) = ln(x + 1) + x^2):

    Estas funciones a suelen presentan comportamientos complejos. Para lr altos puede converger muy rápido al mínimo global, pero para lr bajos puede tomar mas de 100 interacciones y puede no ser exacto y con una convergencia más lenta.

    Las funciones no polinomiales pueden ser más propensas a la inestabilidad o desbordamiento si no se eligen bien los parámetros. una tasa de aprendizaje demasiado alta que hace que el algoritmo salte fuera de la región de interés o diverja

    Las funciones con logaritmos o exponenciales pueden tener regiones de no convexidad, así como asíntotas o puntos donde la derivada no está definida Esto puede hacer que el algoritmo se vaya a regiones no deseadas o que el algoritmo falle en encontrar un mínimo válido si x se acerca a un punto singular, esto se observa en algunos casos en nuestras gráficas

Influencia de las características de las funciones en la dificultad para encontrar mínimos

1.	Si una función es convexa, cualquier mínimo local es también el mínimo global. Esto hace que el descenso del gradiente sea muy efectivo, ya que garantiza la convergencia al mínimo global

    Muchas funciones tienen múltiples mínimos locales y/o máximos locales, lo que crea una optimización compleja. El descenso del gradiente puede quedar atrapado en cualquiera de estos mínimos locales, donde el punto de partida x0 determina cuál mínimo local se encuentra.
    
    Las funciones con derivadas continuas y bien definidas (como los polinomios) permiten que el gradiente sea una buena indicación de la dirección más empinada hacia el mínimo. Esto hace que el descenso del gradiente sea más estable y predecible.

    Las Funciones logarítmicas generalmente tienen un dominio restringido y la derivada se puede acercar a infinito. Si el algoritmo intenta moverse a través o hacia esta región, puede generar valores muy grandes o NaN/inf, llevando a la divergencia o a resultados incorrectos. Esto aumenta enormemente la dificultad.

    Si los gradientes son muy grandes (como puede ocurrir en funciones exponenciales o logarítmicas en ciertas regiones), una learning_rate incluso moderada puede causar saltos excesivos y divergencia. Por otro lado, si los gradientes son muy pequeños, la convergencia puede ser extremadamente lenta. La función h2(x) muestra una gran sensibilidad a lr, donde un lr alto lleva a divergencia, mientras que lr más pequeños parecen llevar a mínimos más razonables (aunque aún pueden ser mínimos locales).

En resumen, las funciones polinomiales (especialmente las de bajo grado o en regiones convexas) suelen ser más 'amigables' para el descenso del gradiente. Las funciones no polinomiales, debido a su mayor complejidad, no convexidad, posibles singularidades, y comportamientos de gradiente más extremos, presentan una mayor dificultad y requieren un análisis más cuidadoso de la tasa de aprendizaje y el punto de inicio para lograr una convergencia efectiva y estable.
