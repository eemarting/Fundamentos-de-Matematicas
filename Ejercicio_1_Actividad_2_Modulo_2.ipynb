Implemente el algoritmo de Descenso del Gradiente desde cero y encuentre los mínimos locales
del siguiente conjunto de funciones polinomiales y trigonométricas.



Instrucciones:

1. Implementación del Algoritmo:
a. Defina una función gradient_descent que acepte como argumentos la función, su derivada, un punto inicial x_start, una tasa de aprendizaje lr, y un número máximo de iteraciones max_iters.
b. Esta función debe retornar la trayectoria de puntos por los que pasó el algoritmo y el último punto obtenido.

2. Aplicación del Algoritmo:
a. Aplique el algoritmo a cada una de las funciones anteriormente indicadas con diferentes puntos iniciales y tasas de aprendizaje.
b. Documente los mínimos que encuentres.

3. Visualización del Algoritmo:
a. Utilizando matplotlib grafique cada función y el camino que sigue el algoritmo de descenso del gradiente.


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# ==========================================
# 1. Función del Algoritmo Base
# ==========================================
def gradient_descent(func, deriv, x_start, lr, max_iters=100, tol=1e-6):
    """
    Ejecuta el algoritmo de descenso del gradiente.
    """
    x = x_start
    history_x = [x]

    for i in range(max_iters):
        grad = deriv(x)
        # Control de desbordamiento (overflow)
        if abs(grad) > 1e4:
            break

        next_x = x - lr * grad
        history_x.append(next_x)

        if abs(next_x - x) < tol:
            x = next_x
            break

        x = next_x

    return history_x, x

# ==========================================
# 2. Definición Simbólica y Preparación
# ==========================================
x_sym = sp.symbols('x')

experimentos = [
    {
        "expr": x_sym**3 + 3*x_sym**2 - 2,
        "name": "a) f(x) = x³ + 3x² - 2",
        "x_starts": [-0.5, 0.5, 1.0],
        "lrs": [0.01, 0.05, 0.1],
        "x_range": (-2.5, 1.5)
    },
    {
        "expr": x_sym**2 * sp.sin(x_sym) + x_sym,
        "name": "b) f(x) = x² sin(x) + x",
        "x_starts": [-2.0, 1.0, 4.0],
        "lrs": [0.01, 0.05, 0.1],
        "x_range": (-6.0, 6.0)
    },
    {
        "expr": 4*x_sym**2 - 10*x_sym + 4,
        "name": "c) f(x) = 4x² - 10x + 4",
        "x_starts": [-1.0, 0.0, 3.0],
        "lrs": [0.01, 0.05, 0.1],
        "x_range": (-1.5, 4.0)
    },
    {
        "expr": x_sym**4 - 4*x_sym**3 + 4,
        "name": "d) f(x) = x⁴ - 4x³ + 4",
        "x_starts": [-1.0, 1.0, 4.0],
        "lrs": [0.001, 0.01, 0.02],
        "x_range": (-2.0, 4.5)
    }
]

# ==========================================
# 3. Ciclos de Ejecución y Graficación (Matriz 3x3)
# ==========================================
for exp in experimentos:
    expr = exp["expr"]
    name = exp["name"]
    x_starts = exp["x_starts"]
    lrs = exp["lrs"]
    x_min, x_max = exp["x_range"]

    # Calcular y mostrar la derivada
    df_expr = sp.diff(expr, x_sym)
    print("=" * 70)
    print(f"Función:  {name}")
    print(f"Derivada: f'(x) = {df_expr}")
    print("=" * 70)

    # Convertir a funciones numéricas de NumPy
    func_num = sp.lambdify(x_sym, expr, modules=['numpy'])
    deriv_num = sp.lambdify(x_sym, df_expr, modules=['numpy'])

    # Crear una matriz de subgráficos de 3 filas x 3 columnas
    fig, axes = plt.subplots(3, 3, figsize=(18, 16))
    fig.suptitle(f"{name}\nDerivada: f'(x) = {df_expr}", fontsize=16, fontweight='bold', y=0.98)

    x_vals = np.linspace(x_min, x_max, 400)
    y_vals = func_num(x_vals)

    # Bucle por filas (Puntos iniciales)
    for i, x_start in enumerate(x_starts):
        # Bucle por columnas (Tasas de aprendizaje)
        for j, lr in enumerate(lrs):
            ax = axes[i, j]

            # Dibujar la función base
            ax.plot(x_vals, y_vals, label='f(x)', color='blue', linewidth=2, alpha=0.3)

            # Ejecutar el algoritmo
            hist_x, min_x = gradient_descent(func_num, deriv_num, x_start, lr)
            hist_y = [func_num(x) for x in hist_x]
            iters = len(hist_x) - 1

            # Imprimir en consola para seguimiento
            print(f"x₀ = {x_start:>4} | LR = {lr:<5} | Iters: {iters:<3} | Mínimo en x = {min_x:.4f}")

            # Dibujar la trayectoria del algoritmo
            ax.plot(hist_x, hist_y, marker='o', markersize=4, linestyle='-',
                    color='red', alpha=0.8, label='Trayectoria')
            ax.plot(min_x, func_num(min_x), marker='*', markersize=10, color='black', label='Mínimo')

            # Configuración del recuadro (AQUÍ SE AGREGARON LAS ITERACIONES)
            ax.set_title(f"Inicio: x₀ = {x_start}  |  lr = {lr}  |  Iters: {iters}", fontsize=12)
            ax.set_xlabel("x")
            ax.set_ylabel("f(x)")
            ax.grid(True, linestyle='--', alpha=0.6)

            # Mostrar la leyenda solo en el primer gráfico para no saturar
            if i == 0 and j == 0:
                ax.legend()

    # Ajustar el espaciado
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
    print("\n")
```

1.	¿Cómo afecta la tasa de aprendizaje a la convergencia del algoritmo?

En los ejercicios anteriores vemos que la tasa de aprendizaje (LR) tiene impacto  en la velocidad (No. De Interacciones) y la de convergencia del algoritmo.

•	Cuando la tasa de aprendizaje es baja, es seguro pero el progreso es lento y requiere mayor número de interacciones para alcanzar la convergencia. Si observamos la función a y c, donde LR = 0.01 se ve que requiere 100 iteraciones o más, lo que indica que alcanzó el max_iters predefinido sin cumplir la tolerancia, o que su avance fue muy lento.
•	Cuando la tasa de aprendizaje es alta, Pueden acelerar la convergencia, como se ve en la función a y c donde LR = 0.1 converge en muy pocas iteraciones, Sin embargo, en funciones más complejas como b o d, una tasa de aprendizaje alta puede causar inestabilidad, oscilaciones o incluso la divergencia del algoritmo, o llevarlo a un mínimo diferente.
•	Cuando la tasa de aprendizaje es moderada (ni baja ni alta) Por lo general, logran un equilibrio, haciendo que el algoritmo converja más rápido (menor número de iteraciones) sin oscilar ni divergir. Esto se observa en casi todos los ejercicios, donde LR = 0.05 o 0.1 reduce significativamente las iteraciones en comparación con LR = 0.01.


Podemos concluir que, una tasa de aprendizaje óptima permite una convergencia rápida y estable, mientras que una muy baja es lenta y una muy alta puede ser inestable.


2.	¿Qué impacto tiene el punto inicial en la localización de mínimos?

El punto inicial (x₀) es muy importante, principalmente en funciones no convexas o con múltiples mínimos locales:

•	Funciones convexas o con un único mínimo global: si observamos el ejercicio con f(x) = 4x² - 10x + 4): El algoritmo tiende a converger al mismo mínimo global y generalmente tienen un solo mínimo local que es el global, independientemente del punto inicial. Esto debido a que la curvatura de la función siempre lleva hacia el mismo valle.

•	Funciones no convexas o con múltiples mínimos, locales: si observamos f(x) = x² sin(x) + x y d) y f(x) = x⁴ - 4x³ + 4):
El punto inicial puede determinar a cuál mínimo local (o global, si tiene suerte) convergerá el algoritmo, esto se puede observar en la función b. y esto nos puede indicar que diferentes puntos de partida nos pueden llevar al algoritmo a diferentes valles.

Es importante considerar o tener en cuenta que el algoritmo de descenso de gradiente es un método local; por lo tanto, su solución depende en gran medida de dónde comienza la búsqueda.


3.	¿En qué funciones el algoritmo enfrenta más dificultades para encontrar el mínimo y por qué?

El algoritmo enfrenta más dificultades en las funciones b) f(x) = x² sin(x) + x y d) f(x) = x⁴ - 4x³ + 4, por lo siguiente:

•	Múltiples Mínimos Locales: Ambas funciones son no convexas y tienen varios mínimos locales. El algoritmo de descenso de gradiente, por su naturaleza, converge al mínimo local más cercano al punto de inicio. Por esta razón puede encontrar diferentes mínimos locales dependiendo del punto de inicio y no forzosamente puede ser el mínimo global. En la función b), se pueden observar diferentes mínimos locales por el punto de inicio.

Si observamos b) f(x) = x² sin(x) + x, donde sin(x) genera oscilaciones que crean varios picos y valles, se dificulta identificar el mínimo global. El algoritmo puede quedar en cualquier valle.

 Si observamos d) f(x) = x⁴ - 4x³ + 4, los términos (x⁴) y 4x³ pueden generar curvas pronunciadas y varios cambios de sentido que generan más de un mínimo local.

Además, las derivadas de funciones con potencias altas pueden crecer rápidamente, lo que requiere tasas de aprendizaje más pequeñas para evitar el desbordamiento o la divergencia, como se observa en las lrs más pequeñas sugeridas para esta función.

Si vemos, la función c) f(x) = 4x² - 10x + 4 es una parábola (convexa), lo que significa que solo tiene un mínimo global, haciendo que el descenso de gradiente sea efectivo y converge consistentemente al mismo punto.
